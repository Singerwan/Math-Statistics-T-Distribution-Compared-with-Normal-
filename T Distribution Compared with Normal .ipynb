```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
dof1=np.arange(12,15)
size1=100000
plt.figure(figsize=(12,6))
yincre=0.065
colors=["red","blue","green"]
for i in range(len(dof1)):
    data=np.random.standard_t(df=dof1[i],size=size1)
    sns.kdeplot(data,color=colors[i],label="T Distribution dof :{}".format(dof1[i]))
    plt.annotate("T Distribution \nDegrees of Freedom:   {}".format(dof1[i]),
                 xy=(2,0.2+yincre*i),color=colors[i],size=18)
    
x_n=np.random.randn(size1)
sns.kdeplot(x_n,lw=3,ls="--",color="gold",alpha=0.6,label="Normal Distrubtion")
plt.legend(loc=(0.05,0.8))
plt.annotate("As the degrees of freedom \napproaches infinity , \nthe t distribution is \napproximating \na normal distribution",
             xytext=(-7,0.15),size=15,color="indigo",
             bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor="cyan",edgecolor="blue") ,
             xy=(-3,0.01),
             arrowprops=dict(arrowstyle="fancy",facecolor="red",edgecolor="green",
                             connectionstyle="arc3,rad=0.3")
            )
plt.axvline(0,0,0.9,ls="--",lw=4,color="black")
plt.savefig("we.pdf")
plt.show()
```
